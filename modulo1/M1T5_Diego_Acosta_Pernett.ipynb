```python
#1.Listas y Tuplas
#Crea una lista y una tupla (≥ 3 strings) de temática libre y realiza:
#-Muestra la lista y la tupla.
lista = [1, 2, 3, 4, 5, 6]
tupla = ("enero", "febrero", "marzo", "abril", "mayo", "junio")

#-Muestra el segundo elemento de la lista y el penúltimo de la tupla.
print(lista[1])
print(tupla[4])

#-Modifica (si se puede) algún elemento de la lista y de la tupla y muestra el resultado.
lista[1] = 10
# tupla[4] = "julio"  Las tuplas son inmutables
print(lista)
print(tupla)

#-Muestra el tamaño de la lista y de la tupla con len().
print(len(lista))
print(len(tupla))

#-Realiza una búsqueda de un elemento con el operador in y muestra si devuelve True o False.
print(10 in lista)
print("julio" in tupla)

#-Añade (si se puede) un elemento a la lista y a la tupla y verifica el cambio.
lista.append(7)
# tupla.append("julio")  Las tuplas son inmutables
print(lista)
print(tupla)

#-Borra o elimina (si se puede) el contenido de la lista y de la tupla y verifica el cambio.
lista.clear()
# tupla.clear()  Las tuplas son inmutables
print(lista)
print(tupla)

```

```python
#2. Sets y Diccionarios
#Crea un set (≥ 3 strings) y un diccionario (≥ 3 pares clave:valor) y realiza:
#-Muestra el set y el diccionario.
persona = {"nombre": "Diego", "apellido": "Acosta", "edad": 19}
gente = {"Juan", "Pedro", "Maria"}

#-Muestra (si se puede) el segundo elemento del set y el valor del primer par del diccionario.
gente = list(gente)  # Convertimos el set a lista para acceder por índice
print(gente[1])

print(persona["nombre"])

#-Modifica (si se puede) algún elemento y muestra el resultado.
persona["edad"] = 20  # Modificamos el valor de la clave 'edad'
gente = {"Juan", "Pedro", "Carlos"}

print(persona)
print(gente)

#-Muestra el tamaño de ambos con len().
print(len(persona))
print(len(gente))

#-Haz una búsqueda en ambos (con in para set y in/.get() para diccionario).
print("Juan" in gente)
print("nombre" in persona)
print(persona.get("nombre"))

#Añade (si se puede) un elemento al set (.add()) y un par clave–valor al diccionario (dic["clave"] = valor).
gente.add("Luis")
persona["ciudad"] = "Granada"

print(gente)
print(persona)

#Borra o elimina (si se puede) el contenido de ambos (.clear()).
gente.clear()
persona.clear()
print(gente)
print(persona)
```

```python
#3. Suma de elementos en una lista
#Pide al usuario tres números float, guárdalos en una lista (sin bucles) y calcula la suma total. Guarda el resultado en sumatorio.
#Pista: usa sum(lista).
input1 = float(input("Introduce el primer número: "))
input2 = float(input("Introduce el segundo número: "))
input3 = float(input("Introduce el tercer número: "))
numeros = [input1, input2, input3]
sumatorio = sum(numeros)
print("La suma total es:", sumatorio)
```

```python
#4. Media aritmética dinámica
#Calcula la media aritmética de los elementos de la lista anterior. Sabemos que el número de elementos introducidos es tres, 
#pero en otra ocasión podría ser diferente; por lo tanto, no puedes usar el número 3 literalmente.
import builtins
valores = [float(x) for x in builtins.input("Introduce los valores separados por comas: ").split(",")]
media = sum(valores) / len(valores)
print("La media aritmética es:", media)
```

```python
#5. Lista de la compra
#Crea una lista con al menos 6 productos: productos = ["pan", "leche", ...].
productos = ["pan", "leche", "huevos", "queso", "fruta", "verdura"]
#Muestra el segundo y el último producto.
print("Segundo producto:", productos[1])
print("Último producto:", productos[-1])

#Modifica el tercer producto y muestra la lista actualizada.
productos[2] = "yogur"
print("Lista actualizada:", productos)

#Intercambia el primero y el último y muestra el resultado.
productos[0], productos[-1] = productos[-1], productos[0]
print("Lista con intercambio:", productos)

#Extrae con slicing:
slicing = productos[1:4]
print("Slicing de productos:", slicing)
#Tres primeros
tres_primeros = productos[:3]
print("Tres primeros:", tres_primeros)

#Tres últimos
tres_últimos = productos[-3:]
print("Tres últimos:", tres_últimos)

#Posiciones pares
posiciones_pares = productos[::2]
print("Posiciones pares:", posiciones_pares)
```

```python
#6. Formateo de cadenas invertidas
#Supón que obtienes una cadena de texto corrupta (al revés). La cadena contiene nombre y año de nacimiento. Debes formatearla para obtener:
#Empleado Nombre Apellido nacido en el año Año
#Cadena de prueba: cadena = "zereP solraC, 5891"
#Pista: invierte con cadena[::-1] y luego separa/da formato.
cadena = "zereP solraC, 5891"
cadena_invertida = cadena[::-1]
año, nombre = [parte.strip() for parte in cadena_invertida.split(",", 1)]
print(f"Empleado {nombre} nacido en el año {año}")
```

```python
#7. Operaciones con conjuntos
#Realiza las siguientes operaciones sobre dos conjuntos:
#Crea un primer conjunto llamado mamiferos con: Gato, León, Caballo, Ballena y Tortuga.
mamiferos = {"Gato", "León", "Caballo", "Ballena", "Tortuga"}

#Crea un segundo conjunto llamado peces con: Ballena y Sardina.
peces = {"Ballena", "Sardina"}
#Borra la Tortuga del conjunto mamiferos.
mamiferos.remove("Tortuga")

#Añade el Pez Payaso al conjunto peces.
peces.add("Pez Payaso")

#Imprime los conjuntos por pantalla.
print("Mamíferos:", mamiferos)
print("Peces:", peces)

#Imprime por pantalla los elementos que se encuentren en ambos conjuntos, es decir, en mamiferos y peces.
#Pistas: mamiferos.remove("Tortuga") (o discard), peces.add("Pez Payaso"); intersección: mamiferos & peces o mamiferos.intersection(peces).
interseccion = mamiferos.intersection(peces)
print("Elementos en ambos conjuntos:", interseccion)
```

```python
#8. Comparación de alumnos
#Crea conjuntos para alumnos de Python e IA y obtén:
#Alumnos en ambos cursos.
alumnos_python = {"Diego", "Ana", "Luis", "Marta"}
alumnos_ia = {"Luis", "Marta", "Carlos", "Lucía"}

#Alumnos exclusivos de Python.
alumnos_exclusivos_python = alumnos_python - alumnos_ia

#Alumnos exclusivos de IA.
alumnos_exclusivos_ia = alumnos_ia - alumnos_python

#Convierte cada resultado a lista y muéstralo.
#Pistas: intersección &; diferencia -; conversión list(conjunto).
print("Alumnos en ambos cursos:", list(alumnos_python & alumnos_ia))
print("Alumnos exclusivos de Python:", list(alumnos_exclusivos_python))
print("Alumnos exclusivos de IA:", list(alumnos_exclusivos_ia))
```

```python
#9. Depurando duplicados
#Dada una lista con elementos repetidos:
#Genera una nueva lista con elementos únicos, eliminando repeticiones automáticamente.
#Guarda el resultado y muéstralo por pantalla.
#Pistas: elimina duplicados con list(set(lista)); si quieres mantener orden, puedes usar list(dict.fromkeys(lista)).ç
lista_con_duplicados = [1, 2, 3, 2, 4, 1, 5, 3]
lista_sin_duplicados = list(set(lista_con_duplicados))
print("Lista sin duplicados:", lista_sin_duplicados)
```

```python
#10. Gestión de tareas con un diccionario
#En este ejercicio, vamos a crear un gestor de tareas donde podremos añadir información progresivamente a un diccionario.
#Contamos con un diccionario con la estructura predefinida:
#"pendientes" → Lista vacía de tareas por hacer.
#"en_progreso" → Lista vacía de tareas que están en marcha.
#"completadas" → Lista vacía de tareas finalizadas.
#Se pedirá al usuario que introduzca el nombre de una tarea.
#Se pedirá al usuario que indique en qué estado debe añadirse: "pendientes", "en_progreso" o "completadas". 
#La tarea se añadirá a la lista correspondiente dentro del diccionario.
#Se mostrará el diccionario actualizado con los cambios.

tareas = {
    "pendientes": [],
    "en_progreso": [],
    "completadas": []
}

while True:
    nombre_tarea = builtins.input("\nIntroduce el nombre de la tarea (o 'salir' para terminar): ")
    if nombre_tarea.strip().lower() == "salir":
        break

    estado = builtins.input("¿En qué estado debe añadirse? (pendientes / en_progreso / completadas): ").strip().lower()

    if estado in tareas:
        tareas[estado].append(nombre_tarea)
        print("\nDiccionario actualizado:")
        print(tareas)
    else:
        print("Estado no válido. Debe ser: pendientes, en_progreso o completadas.")

print("\nResultado final:")
print(tareas)
```
