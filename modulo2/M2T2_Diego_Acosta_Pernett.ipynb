```python
#1. Orden de los números
#Lee tres números por teclado.
#Evalúa si los números están en orden ascendente o no.
#Si el primer número introducido es 0, muestra un mensaje de error.
#Nota: No uses listas en este ejercicio.

numeros = 1, 2, 3

if numeros[0] == 0:
    print("Error: el primer número no puede ser 0.")
elif numeros[0] < numeros[1] < numeros[2]:
    print("Los números están en orden ascendente.")
else:
    print("Los números no están en orden ascendente.")
```

```python
#2. Orden numérico con listas
#Repite el ejercicio anterior, pero introduciendo los números en una lista.

numeros = [1, 2, 3]

if numeros[0] == 0:
    print("Error: el primer número no puede ser 0.")
elif numeros[0] < numeros[1] < numeros[2]:
    print("Los números están en orden ascendente.")
else:
    print("Los números no están en orden ascendente.")
```

```python
#3. Sumatorio de los primeros N números
#Realiza un programa que sume los primeros N números empezando en 0.
#N será un número límite introducido por el usuario.
#Ejemplo: con N = 5 → sumatorio = 0 + 1 + 2 + 3 + 4

N = int(input("Introduce el número límite: "))
sumatorio = 0

for i in range(N):
    sumatorio += i

print(f"El sumatorio de los primeros {N} números es: {sumatorio}")
```

```python
#4. Contador de letras "a"
#Lee letras una por una y cuenta cuántas veces se introduce la letra 'a'.
#El programa termina cuando el usuario introduce el carácter '.'.
#Al finalizar, muestra el número total de veces que se ha introducido la letra 'a'.

contador_a = 0
letra = input("Introduce una letra (o '.' para terminar): ") # Introducir "a" y darle al enter para seguir contando o introducir "." y darle al enter para finalizar el programa.

while letra != '.':
    if letra == 'a':
        contador_a += 1
    letra = input("Introduce una letra (o '.' para terminar): ")

print(f"El número total de veces que se ha introducido la letra 'a' es: {contador_a}")
```

```python
#5. Contador de letras "a" en palabras
#Versión similar al ejercicio anterior, pero leyendo palabras completas.
#Cuenta cuántas veces aparece la letra 'a' en total.
#El programa finaliza cuando se introduce '.'.
#Muestra el total de apariciones de la letra 'a'.

contador_a = 0
palabra = input("Introduce una palabra (o '.' para terminar): ")

while palabra != '.':
    contador_a += palabra.count('a')
    palabra = input("Introduce una palabra (o '.' para terminar): ")

print(f"El número total de veces que se ha introducido la letra 'a' es: {contador_a}")
```

```python
#6. Sumatorio de números pares decrecientes
#Calcula el sumatorio de los números pares comprendidos entre 1000 y 73 (en orden descendente).
#Implementa la solución usando un bucle for o while.

sumatorio = 0

while True:
    for i in range(1000, 72, -1):
        if i % 2 == 0:
            sumatorio += i
    break

print(f"El sumatorio de los números pares comprendidos entre 1000 y 73 es: {sumatorio}")
```

```python
#7. Lista de números: análisis y operaciones
#Solicita al usuario cuántos valores va a introducir.
#Lee todos esos números, guárdalos en una lista y realiza las siguientes operaciones:
#Ordena la lista de menor a mayor.
#Calcula el promedio.
#Identifica el valor máximo y mínimo, y la posición en la que se encuentran.

solicitud = int(input("¿Cuántos valores vas a introducir? "))
numeros = []

for _ in range(solicitud):
    numero = int(input("Introduce un número: "))
    numeros.append(numero)

numeros.sort()
promedio = sum(numeros) / len(numeros)
maximo = max(numeros)
minimo = min(numeros)
pos_max = numeros.index(maximo)
pos_min = numeros.index(minimo)

print(f"Lista ordenada: {numeros}")
print(f"Promedio: {promedio}")
print(f"Valor máximo: {maximo} (posición: {pos_max})")
print(f"Valor mínimo: {minimo} (posición: {pos_min})")
```

```python
#8. Menú compra-venta
#Crea un menú con la siguiente estructura:

#MENU COMPRA-VENTA
#1. Comprar
#2. Vender
#3. Alquilar
#4. Mostrar listas de datos
#5. Salir

#Si elige 1: pregunta qué desea comprar y guárdalo en una lista de compras.
#Si elige 2: pregunta qué desea vender y guárdalo en una lista de ventas.
#Si elige 3: pregunta qué desea alquilar y guárdalo en una lista de alquileres.
#Si elige 4: muestra todas las listas.
#Si elige 5: finaliza el programa.

compras = []
ventas = []
alquileres = []

while True:
    print("\nMENU COMPRA-VENTA")
    print("1. Comprar")
    print("2. Vender")
    print("3. Alquilar")
    print("4. Mostrar listas de datos")
    print("5. Salir")

    opcion = input("Elige una opción (1-5): ")

    if opcion == '1':
        item = input("¿Qué deseas comprar? ")
        compras.append(item)
        print(f"{item} ha sido añadido a la lista de compras.")
    elif opcion == '2':
        item = input("¿Qué deseas vender? ")
        ventas.append(item)
        print(f"{item} ha sido añadido a la lista de ventas.")
    elif opcion == '3':
        item = input("¿Qué deseas alquilar? ")
        alquileres.append(item)
        print(f"{item} ha sido añadido a la lista de alquileres.")
    elif opcion == '4':
        print("\nLista de compras:", compras)
        print("Lista de ventas:", ventas)
        print("Lista de alquileres:", alquileres)
    elif opcion == '5':
        print("Saliendo del programa...")
        break
    else:
        print("Opción no válida. Por favor, elige una opción del 1 al 5.")
```

```python
#9. Programa de login con reintentos
#Solicita una contraseña y guárdala en una variable.
#Pide al usuario que introduzca la contraseña para iniciar sesión.
#Si la contraseña es correcta, muestra "Login correcto".
#Si es incorrecta, muestra "Login fallido. Intento X", donde X aumenta hasta 5.
#Si llega a 5 intentos fallidos, el programa se reinicia desde el paso 1.

contraseña = input("Introduce una contraseña para guardarla: ")

for intento in range(1, 6):
    contraseña_ingresada = input("Introduce la contraseña para iniciar sesión: ")
    if contraseña_ingresada == contraseña:
        print("Login correcto")
        break
    else:
        print(f"Login fallido. Intento {intento}")
else:
    print("Has alcanzado el número máximo de intentos fallidos. El programa se reinicia.")
```

```python
#10. Verificar aprobados y suspendidos con un diccionario
#Dado un diccionario con nombres como claves y calificaciones como valores, recórrelo y muestra:
#"Aprobado" si la nota es mayor o igual a 5.
#"Suspendido" si la nota es menor que 5.
#Ejemplo: notas = {"Ana": 8, "Luis": 4, "Marta": 6, "Pedro": 3}

notas = {"Ana": 8, "Luis": 4, "Marta": 6, "Pedro": 3}

for nombre, calificacion in notas.items():
    if calificacion >= 5:
        print(f"{nombre}: Aprobado")
    else:
        print(f"{nombre}: Suspendido")
```
