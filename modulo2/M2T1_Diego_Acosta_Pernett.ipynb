```python
#2. Índices de cadenas
#Crea texto = "Python".
#Imprime la primera letra usando índices.
#Imprime la última letra usando índices negativos.

texto = "Python"
print("La primera letra es:", texto[0])
print("La última letra es:", texto[-1])
```

```python
#3. Extraer partes de una dirección (slicing)
#Con direccion = "Calle Falsa, 123, Springfield":

#Extrae e imprime solo el nombre de la calle.
#Extrae e imprime solo el número.
#Extrae e imprime solo la ciudad.

direccion = "Calle Falsa, 123, Springfield"
nombre_calle = direccion.split(",")[0]
numero = direccion.split(",")[1].strip()
ciudad = direccion.split(",")[2].strip()
print("Nombre de la calle:", nombre_calle)
print("Número:", numero)
print("Ciudad:", ciudad)
```

```python
#4. Identificador único con slicing y concatenación
#Define nombre, apellido y año_nacimiento (string).
#Extrae las 3 primeras letras de nombre y apellido, y los 2 últimos dígitos del año.
#Concatena para formar un identificador (p. ej. CriRod89) e imprímelo.

nombre = "Cristian"
apellido = "Rodriguez"
año_nacimiento = "1989"
identificador = nombre[:3] + apellido[:3] + año_nacimiento[-2:]
print("Identificador único:", identificador)
```

```python
#5. Inmutabilidad: modificar una cadena
#Crea palabra = "robot".
#Intenta palabra[0] = "R" y observa el error.
#Crea palabra_modificada con slicing y concatenación para obtener "Robot".
#Imprime el resultado.

palabra = "robot"
# Intenta modificar la primera letra (esto generará un error)
# palabra[0] = "R"  # Esto generará un error porque las cadenas son inmutables
palabra_modificada = "R" + palabra[1:]
print("Palabra modificada:", palabra_modificada)
```

```python
#6. Analizador de palabras
#Dados: texto = "Python es POTENTE. Aprender python es útil. PYTHON abre puertas." y objetivo = "Python".
#Convierte ambos a minúsculas para búsqueda insensible a mayúsculas.
#Cuenta apariciones como palabra exacta (con split()) y como subcadena (con .count()).
#Muestra ambos totales.
#Extra: genera una versión del texto remarcando las apariciones de la palabra objetivo en MAYÚSCULAS.
#Salida esperada (aproximada): totales de coincidencias y texto resaltado, por ejemplo: PYTHON es POTENTE...

texto = "Python es POTENTE. Aprender python es útil. PYTHON abre puertas."
objetivo = "Python"

# Convertir ambos a minúsculas para búsqueda insensible a mayúsculas
texto_lower = texto.lower()

# Contar apariciones como palabra exacta
palabras = texto_lower.split()
conteo_exacto = sum(1 for palabra in palabras if palabra.strip('.,') == objetivo.lower())

# Contar apariciones como subcadena
conteo_subcadena = texto_lower.count(objetivo.lower())

# Generar versión del texto remarcando las apariciones de la palabra objetivo en MAYÚSCULAS
texto_resaltado = texto.replace(objetivo, objetivo.upper()).replace(objetivo.lower(), objetivo.upper())

print("Conteo exacto de la palabra:", conteo_exacto)
print("Conteo como subcadena:", conteo_subcadena)
print("Texto resaltado:", texto_resaltado)
```
